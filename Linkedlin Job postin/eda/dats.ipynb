```python
columns_to_drop = [
    'closed_time',           # 99.56% missing
    'skills_desc',           # 99.47% missing
    'compensation_type',     # Only one value (BASE_SALARY)
    'normalized_salary'      # 70.79% missing, redundant with min/max
]
df = df.drop(columns=columns_to_drop)  
```

```python
# Check for exact duplicates (already done - NONE found ✓)
df = df.drop_duplicates()

# Optional: Check for near-duplicates by job_id and title
df = df.drop_duplicates(subset=['job_id'], keep='first')
```

```python
# Fill missing company names
df['company_name'] = df['company_name'].fillna('Unknown Company')

# Fill missing company_id with -1 (indicator for unknown)
df['company_id'] = df['company_id'].fillna(-1).astype(int)
```

```python
# Replace formatted_work_type with work_type if more complete
# Check consistency: they should match (formatted = pretty version)
df['work_type'] = df['work_type'].str.upper()

# Standardize values
work_type_map = {
    'FULL_TIME': 'Full-time',
    'PART_TIME': 'Part-time',
    'CONTRACT': 'Contract',
    'INTERNSHIP': 'Internship',
    'TEMPORARY': 'Temporary'
}
df['work_type'] = df['work_type'].map(work_type_map)

# Drop redundant formatted_work_type
df = df.drop(columns=['formatted_work_type'])
```

```python
# Handle missing salary values
# Strategy: Keep nulls for now (indicate missing data)
# Later: Use pay_period to infer missing values if possible

# Remove outliers (data entry errors)
# Assuming min_salary < max_salary
df_clean = df[(df['min_salary'] <= df['max_salary']) | df['min_salary'].isnull()]

# Remove extremely low/high salaries (likely errors)
df_clean = df_clean[
    (df_clean['min_salary'] > 7) | (df_clean['min_salary'].isnull())
]  # $7/hour is too low

# Create salary range (for analysis)
df_clean['salary_range'] = df_clean['max_salary'] - df_clean['min_salary']

# Standardize currency to USD
df_clean.loc[df_clean['currency'] == 'CAD', 'min_salary'] *= 0.74
df_clean.loc[df_clean['currency'] == 'CAD', 'max_salary'] *= 0.74
df_clean.loc[df_clean['currency'] == 'CAD', 'currency'] = 'USD'
```

```python
# Fill missing pay_period with 'YEARLY' (most common)
df['pay_period'] = df['pay_period'].fillna('YEARLY')

# Standardize values
pay_period_map = {
    'YEARLY': 'Annual',
    'HOURLY': 'Hourly',
    'MONTHLY': 'Monthly',
    'WEEKLY': 'Weekly'
}
df['pay_period'] = df['pay_period'].map(pay_period_map)
```

```python
# Fill missing experience level
df['formatted_experience_level'] = df['formatted_experience_level'].fillna('Not Specified')

# Standardize capitalization
df['formatted_experience_level'] = df['formatted_experience_level'].str.title()

# Create experience level category
level_order = ['Internship', 'Entry level', 'Associate', 'Mid-Senior level', 'Director', 'Executive', 'Not Specified']
df['experience_level_encoded'] = pd.Categorical(
    df['formatted_experience_level'], 
    categories=level_order, 
    ordered=True
)
```

```python
# Fill missing locations with "Remote" or "Not Specified"
df['location'] = df['location'].fillna('Not Specified')

# Extract country code from location if present
df['country'] = df['location'].str.extract(r'([A-Z]{2})$')
df['country'] = df['country'].fillna('US')

# Extract state/province (for US jobs)
df['state'] = df.apply(lambda row: row['location'].split(',')[-2].strip() 
                       if ',' in str(row['location']) else 'Unknown', axis=1)

# Fill zip_code and fips from location if possible
df['zip_code'] = df['zip_code'].fillna(-1).astype(int)
df['fips'] = df['fips'].fillna(-1).astype(int)
```

```python
# remote_allowed is binary but mostly missing
# Replace with 0/1 and fill missing with 0 (assume not remote if not specified)
df['remote_allowed'] = df['remote_allowed'].fillna(0).astype(int)
```

```python
# Standardize application_type values
df['application_type'] = df['application_type'].str.upper()

app_type_map = {
    'OFFSITEAPPLY': 'External Link',
    'COMPLEXONSITEAPPLY': 'Complex Form',
    'SIMPLEONSITEAPPLY': 'Simple Form'
}
df['application_type'] = df['application_type'].map(app_type_map)

# Fill missing with most common
df['application_type'] = df['application_type'].fillna('External Link')

# Validate URLs
df['job_posting_url'] = df['job_posting_url'].fillna('Missing')
df['application_url'] = df['application_url'].fillna(df['job_posting_url'])

# Extract domain from URLs
df['posting_domain'] = df['posting_domain'].fillna('Unknown')
```

```python
# Handle missing values in views and applies
df['views'] = df['views'].fillna(0).astype(int)
df['applies'] = df['applies'].fillna(0).astype(int)

# Calculate engagement rate (prevent division by zero)
df['engagement_rate'] = df.apply(
    lambda row: (row['applies'] / row['views'] * 100) if row['views'] > 0 else 0,
    axis=1
)

# Remove outliers in engagement (99th percentile)
q99 = df['engagement_rate'].quantile(0.99)
df['engagement_rate'] = df['engagement_rate'].clip(upper=q99)
```

```python
# Convert timestamps to datetime
df['original_listed_time'] = pd.to_datetime(df['original_listed_time'], unit='s', errors='coerce')
df['listed_time'] = pd.to_datetime(df['listed_time'], unit='s', errors='coerce')
df['expiry'] = pd.to_datetime(df['expiry'], unit='s', errors='coerce')

# Fill missing dates with median date
median_date = df['original_listed_time'].median()
df['original_listed_time'] = df['original_listed_time'].fillna(median_date)
df['listed_time'] = df['listed_time'].fillna(median_date)

# Create posting duration (days until expiry)
df['posting_duration_days'] = (df['expiry'] - df['original_listed_time']).dt.days
df['posting_duration_days'] = df['posting_duration_days'].fillna(0).astype(int)
```

```python
# Ensure binary encoding
df['sponsored'] = df['sponsored'].astype(int)
assert df['sponsored'].isin([0, 1]).all(), "Sponsored should be 0 or 1"
```

```python
"""
Job Postings Dataset Cleaning Script
Complete step-by-step data cleaning for portfolio analysis
"""

import pandas as pd
import numpy as np
from datetime import datetime
import logging

# Setup logging
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s'
)

class JobPostingsDataCleaner:
    def __init__(self, filepath):
        self.filepath = filepath
        self.df = None
        self.df_original = None
        self.cleaning_log = []
        
    def load_data(self, nrows=None):
        """Load the CSV file"""
        logging.info(f"Loading data from {self.filepath}...")
        self.df = pd.read_csv(self.filepath, nrows=nrows)
        self.df_original = self.df.copy()
        logging.info(f"Loaded {len(self.df):,} rows × {len(self.df.columns)} columns")
        return self
        
    def log_step(self, step_name, rows_before, rows_after=None):
        """Log cleaning steps"""
        if rows_after is None:
            rows_after = len(self.df)
        rows_removed = rows_before - rows_after
        pct_removed = (rows_removed / rows_before * 100) if rows_before > 0 else 0
        
        message = f"✓ {step_name} | Rows: {rows_after:,} | Removed: {rows_removed:,} ({pct_removed:.2f}%)"
        logging.info(message)
        self.cleaning_log.append(message)
        
    def step1_remove_useless_columns(self):
        """STEP 1: Remove columns with >95% missing or no variance"""
        logging.info("\n" + "="*80)
        logging.info("STEP 1: Remove Useless Columns")
        logging.info("="*80)
        
        columns_to_drop = [
            'closed_time',           # 99.56% missing
            'skills_desc',           # 99.47% missing
            'compensation_type',     # Only one value
        ]
        
        rows_before = len(self.df)
        self.df = self.df.drop(columns=columns_to_drop)
        
        logging.info(f"Dropped columns: {columns_to_drop}")
        self.log_step("Remove useless columns", rows_before)
        return self
        
    def step2_handle_duplicates(self):
        """STEP 2: Remove duplicate job postings"""
        logging.info("\n" + "="*80)
        logging.info("STEP 2: Handle Duplicate Job Postings")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        # Check exact duplicates
        exact_dupes = self.df.duplicated().sum()
        logging.info(f"Exact duplicates: {exact_dupes}")
        
        # Check job_id duplicates (keep first)
        job_dupes = self.df['job_id'].duplicated().sum()
        logging.info(f"Duplicate job_ids: {job_dupes}")
        
        # Drop by job_id (keeping first occurrence)
        self.df = self.df.drop_duplicates(subset=['job_id'], keep='first')
        
        self.log_step("Remove duplicates", rows_before)
        return self
        
    def step3_clean_company_info(self):
        """STEP 3: Clean company information"""
        logging.info("\n" + "="*80)
        logging.info("STEP 3: Clean Company Information")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        # Fill missing company names
        missing_company = self.df['company_name'].isnull().sum()
        logging.info(f"Missing company names: {missing_company:,}")
        self.df['company_name'] = self.df['company_name'].fillna('Unknown Company')
        
        # Fill missing company_id
        missing_company_id = self.df['company_id'].isnull().sum()
        logging.info(f"Missing company_ids: {missing_company_id:,}")
        self.df['company_id'] = self.df['company_id'].fillna(-1).astype(int)
        
        self.log_step("Clean company information", rows_before)
        return self
        
    def step4_standardize_work_type(self):
        """STEP 4: Standardize work type"""
        logging.info("\n" + "="*80)
        logging.info("STEP 4: Standardize Work Type")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        # Map work_type to formatted values
        work_type_map = {
            'FULL_TIME': 'Full-time',
            'PART_TIME': 'Part-time',
            'CONTRACT': 'Contract',
            'INTERNSHIP': 'Internship',
            'TEMPORARY': 'Temporary'
        }
        
        self.df['work_type'] = self.df['work_type'].str.upper()
        self.df['work_type'] = self.df['work_type'].map(work_type_map)
        self.df['work_type'] = self.df['work_type'].fillna('Unknown')
        
        logging.info(f"Work type distribution:\n{self.df['work_type'].value_counts()}")
        
        # Drop redundant formatted_work_type
        self.df = self.df.drop(columns=['formatted_work_type'])
        
        self.log_step("Standardize work type", rows_before)
        return self
        
    def step5_clean_salary_data(self):
        """STEP 5: Clean and validate salary data"""
        logging.info("\n" + "="*80)
        logging.info("STEP 5: Clean Salary Data")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        # Basic statistics before cleaning
        logging.info(f"Min salary - Mean: ${self.df['min_salary'].mean():,.2f}, Median: ${self.df['min_salary'].median():,.2f}")
        logging.info(f"Max salary - Mean: ${self.df['max_salary'].mean():,.2f}, Median: ${self.df['max_salary'].median():,.2f}")
        
        # Remove rows where min > max (data entry errors)
        invalid_salary = ((self.df['min_salary'] > self.df['max_salary']) & 
                         (self.df['min_salary'].notna() & self.df['max_salary'].notna())).sum()
        logging.info(f"Invalid salary ranges (min > max): {invalid_salary}")
        
        self.df = self.df[
            (self.df['min_salary'] <= self.df['max_salary']) | 
            (self.df['min_salary'].isnull()) | 
            (self.df['max_salary'].isnull())
        ]
        
        # Create salary range column
        self.df['salary_range'] = self.df['max_salary'] - self.df['min_salary']
        
        # Create average salary
        self.df['avg_salary'] = (self.df['min_salary'] + self.df['max_salary']) / 2
        
        logging.info(f"Salary range distribution:\n{self.df['salary_range'].describe()}")
        
        self.log_step("Clean salary data", rows_before)
        return self
        
    def step6_clean_currency(self):
        """STEP 6: Standardize currency to USD"""
        logging.info("\n" + "="*80)
        logging.info("STEP 6: Standardize Currency")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        logging.info(f"Currency distribution:\n{self.df['currency'].value_counts()}")
        
        # Convert CAD to USD (approximate rate: 1 CAD = 0.74 USD)
        cad_rows = self.df['currency'] == 'CAD'
        cad_count = cad_rows.sum()
        logging.info(f"Converting {cad_count} CAD entries to USD...")
        
        self.df.loc[cad_rows, 'min_salary'] *= 0.74
        self.df.loc[cad_rows, 'max_salary'] *= 0.74
        self.df.loc[cad_rows, 'salary_range'] *= 0.74
        self.df.loc[cad_rows, 'avg_salary'] *= 0.74
        self.df.loc[cad_rows, 'currency'] = 'USD'
        
        self.log_step("Standardize currency", rows_before)
        return self
        
    def step7_clean_pay_period(self):
        """STEP 7: Clean and standardize pay period"""
        logging.info("\n" + "="*80)
        logging.info("STEP 7: Clean Pay Period")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        logging.info(f"Pay period distribution (before):\n{self.df['pay_period'].value_counts(dropna=False)}")
        
        # Fill missing with 'YEARLY' (most common)
        missing_pay_period = self.df['pay_period'].isnull().sum()
        logging.info(f"Missing pay periods: {missing_pay_period:,} -> filling with YEARLY")
        self.df['pay_period'] = self.df['pay_period'].fillna('YEARLY')
        
        # Standardize format
        pay_period_map = {
            'YEARLY': 'Annual',
            'HOURLY': 'Hourly',
            'MONTHLY': 'Monthly',
            'WEEKLY': 'Weekly'
        }
        self.df['pay_period'] = self.df['pay_period'].map(pay_period_map)
        self.df['pay_period'] = self.df['pay_period'].fillna('Annual')
        
        logging.info(f"Pay period distribution (after):\n{self.df['pay_period'].value_counts()}")
        
        self.log_step("Clean pay period", rows_before)
        return self
        
    def step8_clean_experience_level(self):
        """STEP 8: Clean experience level"""
        logging.info("\n" + "="*80)
        logging.info("STEP 8: Clean Experience Level")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        missing_exp = self.df['formatted_experience_level'].isnull().sum()
        logging.info(f"Missing experience levels: {missing_exp:,}")
        
        self.df['formatted_experience_level'] = self.df['formatted_experience_level'].fillna('Not Specified')
        
        # Standardize capitalization
        self.df['formatted_experience_level'] = self.df['formatted_experience_level'].str.title()
        
        logging.info(f"Experience level distribution:\n{self.df['formatted_experience_level'].value_counts()}")
        
        self.log_step("Clean experience level", rows_before)
        return self
        
    def step9_clean_location(self):
        """STEP 9: Clean location data"""
        logging.info("\n" + "="*80)
        logging.info("STEP 9: Clean Location Data")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        missing_location = self.df['location'].isnull().sum()
        logging.info(f"Missing locations: {missing_location:,}")
        
        self.df['location'] = self.df['location'].fillna('Not Specified')
        
        # Extract country (assuming last 2 characters are country code)
        self.df['country'] = self.df['location'].str.extract(r'([A-Z]{2})$')
        self.df['country'] = self.df['country'].fillna('US')
        # Extract state from location
        self.df['state'] = self.df['location'].str.split(',').str[0]
        logging.info(f"Country distribution:\n{self.df['country'].value_counts().head(10)}")
        
        # Fill zip_code and fips with placeholder
        self.df['zip_code'] = self.df['zip_code'].fillna(-1).astype(int)
        self.df['fips'] = self.df['fips'].fillna(-1).astype(int)
        
        self.log_step("Clean location", rows_before)
        return self
        
    def step10_clean_remote_work(self):
        """STEP 10: Clean remote work flag"""
        logging.info("\n" + "="*80)
        logging.info("STEP 10: Clean Remote Work Flag")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        missing_remote = self.df['remote_allowed'].isnull().sum()
        logging.info(f"Missing remote_allowed values: {missing_remote:,}")
        
        # Convert to binary and fill missing with 0
        self.df['remote_allowed'] = self.df['remote_allowed'].fillna(0).astype(int)
        
        logging.info(f"Remote work distribution:\n{self.df['remote_allowed'].value_counts()}")
        
        self.log_step("Clean remote work flag", rows_before)
        return self
        
    def step11_clean_application_data(self):
        """STEP 11: Clean application information"""
        logging.info("\n" + "="*80)
        logging.info("STEP 11: Clean Application Data")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        # Standardize application_type
        logging.info(f"Application type distribution:\n{self.df['application_type'].value_counts(dropna=False)}")
        
        app_type_map = {
            'OFFSITEAPPLY': 'External Link',
            'COMPLEXONSITEAPPLY': 'Complex Form',
            'SIMPLEONSITEAPPLY': 'Simple Form'
        }
        
        self.df['application_type'] = self.df['application_type'].str.upper()
        self.df['application_type'] = self.df['application_type'].map(app_type_map)
        self.df['application_type'] = self.df['application_type'].fillna('External Link')
        
        # Fill application_url with posting_url if missing
        missing_app_url = self.df['application_url'].isnull().sum()
        logging.info(f"Missing application URLs: {missing_app_url:,}")
        
        self.df['application_url'] = self.df['application_url'].fillna(self.df['job_posting_url'])
        
        # Handle missing URLs
        self.df['job_posting_url'] = self.df['job_posting_url'].fillna('Missing')
        self.df['application_url'] = self.df['application_url'].fillna('Missing')
        
        self.log_step("Clean application data", rows_before)
        return self
        
    def step12_clean_engagement_metrics(self):
        """STEP 12: Clean engagement metrics"""
        logging.info("\n" + "="*80)
        logging.info("STEP 12: Clean Engagement Metrics")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        # Fill missing values
        self.df['views'] = self.df['views'].fillna(0).astype(int)
        self.df['applies'] = self.df['applies'].fillna(0).astype(int)
        
        logging.info(f"Views - Mean: {self.df['views'].mean():.1f}, Median: {self.df['views'].median():.1f}")
        logging.info(f"Applies - Mean: {self.df['applies'].mean():.1f}, Median: {self.df['applies'].median():.1f}")
        
        # Calculate engagement rate
        self.df['engagement_rate'] = self.df.apply(
            lambda row: (row['applies'] / row['views'] * 100) if row['views'] > 0 else 0,
            axis=1
        )
        
        # Remove extreme outliers (99th percentile)
        q99 = self.df['engagement_rate'].quantile(0.99)
        logging.info(f"Capping engagement rate at 99th percentile: {q99:.2f}%")
        self.df['engagement_rate'] = self.df['engagement_rate'].clip(upper=q99)
        
        self.log_step("Clean engagement metrics", rows_before)
        return self
        
    def step13_clean_timestamps(self):
        """STEP 13: Clean and standardize timestamps"""
        logging.info("\n" + "="*80)
        logging.info("STEP 13: Clean Timestamps")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        # Convert Unix timestamps to datetime
        logging.info("Converting Unix timestamps to datetime...")
        self.df['original_listed_time'] = pd.to_datetime(
            self.df['original_listed_time'], 
            unit='s', 
            errors='coerce'
        )
        self.df['listed_time'] = pd.to_datetime(
            self.df['listed_time'], 
            unit='s', 
            errors='coerce'
        )
        self.df['expiry'] = pd.to_datetime(
            self.df['expiry'], 
            unit='s', 
            errors='coerce'
        )
        
        # Fill missing dates with median
        median_date = self.df['original_listed_time'].median()
        logging.info(f"Median date: {median_date}")
        
        self.df['original_listed_time'] = self.df['original_listed_time'].fillna(median_date)
        self.df['listed_time'] = self.df['listed_time'].fillna(median_date)
        
        # Create posting duration
        self.df['posting_duration_days'] = (self.df['expiry'] - self.df['original_listed_time']).dt.days
        self.df['posting_duration_days'] = self.df['posting_duration_days'].fillna(0).astype(int)
        
        logging.info(f"Posting duration - Mean: {self.df['posting_duration_days'].mean():.1f} days")
        
        self.log_step("Clean timestamps", rows_before)
        return self
        
    def step14_clean_sponsored(self):
        """STEP 14: Validate sponsored flag"""
        logging.info("\n" + "="*80)
        logging.info("STEP 14: Clean Sponsored Flag")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        self.df['sponsored'] = self.df['sponsored'].astype(int)
        
        assert self.df['sponsored'].isin([0, 1]).all(), "Sponsored should be 0 or 1"
        logging.info(f"Sponsored distribution:\n{self.df['sponsored'].value_counts()}")
        
        self.log_step("Validate sponsored flag", rows_before)
        return self
        
    def step15_remove_incomplete_rows(self):
        """STEP 15: Remove rows with critical missing values"""
        logging.info("\n" + "="*80)
        logging.info("STEP 15: Remove Incomplete Rows")
        logging.info("="*80)
        
        rows_before = len(self.df)
        
        # Critical columns that should always have values
        critical_cols = ['job_id', 'title', 'company_name', 'location']
        
        missing_critical = self.df[critical_cols].isnull().sum()
        logging.info(f"Missing values in critical columns:\n{missing_critical}")
        
        self.df = self.df.dropna(subset=critical_cols)
        
        self.log_step("Remove rows with critical missing values", rows_before)
        return self
        
    def final_validation(self):
        """STEP 16: Final data quality checks"""
        logging.info("\n" + "="*80)
        logging.info("STEP 16: Final Validation & Quality Checks")
        logging.info("="*80)
        
        # Check for remaining nulls
        null_counts = self.df.isnull().sum()
        null_cols = null_counts[null_counts > 0]
        
        if len(null_cols) > 0:
            logging.info(f"Remaining null values:\n{null_cols}")
        else:
            logging.info("✓ No null values remaining!")
        
        # Data type check
        logging.info(f"\nFinal data types:\n{self.df.dtypes}")
        
        # Logical consistency checks
        salary_check = (self.df['min_salary'] <= self.df['max_salary']).all() or self.df['min_salary'].isnull().any()
        assert salary_check, "Invalid salary ranges found"
        logging.info("✓ Salary ranges are valid")
        
        invalid_rows = self.df[self.df['applies'] > self.df['views']]

        logging.info(f"Invalid rows where applies > views: {len(invalid_rows)}")

# Remove invalid rows
        self.df = self.df[
        (self.df['views'] >= self.df['applies']) |
        (self.df['views'].isnull()) |
        (self.df['applies'].isnull())
]

        logging.info("✓ Invalid engagement rows removed")
        
        # Summary statistics
        logging.info(f"\n" + "="*80)
        logging.info(f"FINAL DATASET SUMMARY")
        logging.info(f"="*80)
        logging.info(f"Total rows: {len(self.df):,}")
        logging.info(f"Total columns: {len(self.df.columns)}")
        logging.info(f"Total rows removed: {len(self.df_original) - len(self.df):,}")
        logging.info(f"Rows retained: {len(self.df) / len(self.df_original) * 100:.2f}%")
        
        return self
        
    def run_full_cleaning(self):
        """Execute all cleaning steps"""
        logging.info("\n" + "="*80)
        logging.info("STARTING FULL DATA CLEANING PIPELINE")
        logging.info("="*80 + "\n")
        
        (self.load_data()
         .step1_remove_useless_columns()
         .step2_handle_duplicates()
         .step3_clean_company_info()
         .step4_standardize_work_type()
         .step5_clean_salary_data()
         .step6_clean_currency()
         .step7_clean_pay_period()
         .step8_clean_experience_level()
         .step9_clean_location()
         .step10_clean_remote_work()
         .step11_clean_application_data()
         .step12_clean_engagement_metrics()
         .step13_clean_timestamps()
         .step14_clean_sponsored()
         .step15_remove_incomplete_rows()
         .final_validation())
        
        return self
        
    def save_cleaned_data(self, output_path):
        """Save cleaned dataset"""
        logging.info(f"\nSaving cleaned dataset to {output_path}...")
        self.df.to_csv(output_path, index=False)
        logging.info(f"✓ Saved {len(self.df):,} rows to {output_path}")
        
    def get_summary_report(self):
        """Generate summary report"""
        report = f"""
{'='*80}
DATA CLEANING SUMMARY REPORT
{'='*80}

Original Dataset: {len(self.df_original):,} rows × {len(self.df_original.columns)} columns
Cleaned Dataset:  {len(self.df):,} rows × {len(self.df.columns)} columns

Rows Removed: {len(self.df_original) - len(self.df):,} ({(len(self.df_original) - len(self.df)) / len(self.df_original) * 100:.2f}%)
Rows Retained: {len(self.df):,} ({len(self.df) / len(self.df_original) * 100:.2f}%)

Null Values by Column:
{self.df.isnull().sum().to_string()}

{'='*80}
CLEANING STEPS COMPLETED:
{'='*80}
"""
        for log in self.cleaning_log:
            report += f"\n{log}"
        
        return report


# ============================================================================
# USAGE EXAMPLE
# ============================================================================

if __name__ == "__main__":
    # Initialize cleaner
    cleaner = JobPostingsDataCleaner(r"C:\Users\USER\OneDrive\Desktop\Linkedlin Job postin\data\raw\postings.csv")
    
    # Run full cleaning pipeline
    cleaner.run_full_cleaning()
    
    # Save cleaned data
    cleaner.save_cleaned_data('postings_cleaned.csv')
    
    # Print summary report
    print(cleaner.get_summary_report())
    
    # Optional: Get the cleaned dataframe
    df_clean = cleaner.df
    print("\nCleaned dataset is ready for analysis!")
    print(f"Shape: {df_clean.shape}")
    print(f"\nFirst few rows:\n{df_clean.head()}")
```

```python
"""
Exploratory Data Analysis (EDA) - Job Postings Dataset
Complete analysis for portfolio insights
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import warnings

warnings.filterwarnings('ignore')

# Set style for better visualizations
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 8)

class JobPostingsEDA:
    def __init__(self, df):
        self.df = df
        self.analysis_results = {}
        
    # =========================================================================
    # EXPLORATORY ANALYSIS METHODS
    # =========================================================================
    
    def analyze_missing_values(self):
        """Analyze missing values before cleaning"""
        print("\n" + "="*80)
        print("MISSING VALUE ANALYSIS")
        print("="*80)
        
        missing = self.df.isnull().sum().sort_values(ascending=False)
        missing_pct = (missing / len(self.df) * 100).round(2)
        
        missing_df = pd.DataFrame({
            'Missing_Count': missing,
            'Percentage': missing_pct
        })
        
        print(missing_df[missing_df['Missing_Count'] > 0])
        self.analysis_results['missing_values'] = missing_df
        
        return self
        
    def analyze_salary_distribution(self):
        """Analyze salary data across categories"""
        print("\n" + "="*80)
        print("SALARY DISTRIBUTION ANALYSIS")
        print("="*80)
        
        # Overall stats
        print("\nOverall Salary Statistics:")
        print(f"Min Salary - Mean: ${self.df['min_salary'].mean():,.2f}")
        print(f"Max Salary - Mean: ${self.df['max_salary'].mean():,.2f}")
        print(f"Salary Range: ${self.df['salary_range'].mean():,.2f} average")
        
        # By work type
        print("\nAverage Salary by Work Type:")
        salary_by_work = self.df.groupby('work_type').agg({
            'min_salary': 'mean',
            'max_salary': 'mean',
            'salary_range': 'mean'
        }).round(2)
        print(salary_by_work)
        self.analysis_results['salary_by_work_type'] = salary_by_work
        
        # By experience level
        print("\nAverage Salary by Experience Level:")
        salary_by_exp = self.df.groupby('formatted_experience_level').agg({
            'min_salary': 'mean',
            'max_salary': 'mean',
            'job_id': 'count'
        }).round(2)
        salary_by_exp.columns = ['Min_Salary', 'Max_Salary', 'Count']
        print(salary_by_exp.sort_values('Max_Salary', ascending=False))
        self.analysis_results['salary_by_experience'] = salary_by_exp
        
        # By pay period
        print("\nSalary by Pay Period:")
        print(self.df[self.df['max_salary'].notna()].groupby('pay_period')[['min_salary', 'max_salary']].agg(['mean', 'count']))
        
        return self
        
    def analyze_work_types(self):
        """Analyze distribution of work types"""
        print("\n" + "="*80)
        print("WORK TYPE ANALYSIS")
        print("="*80)
        
        work_dist = self.df['work_type'].value_counts()
        work_pct = (work_dist / len(self.df) * 100).round(2)
        
        work_analysis = pd.DataFrame({
            'Count': work_dist,
            'Percentage': work_pct
        })
        
        print(work_analysis)
        self.analysis_results['work_type_distribution'] = work_analysis
        
        return self
        
    def analyze_experience_levels(self):
        """Analyze distribution of experience levels"""
        print("\n" + "="*80)
        print("EXPERIENCE LEVEL ANALYSIS")
        print("="*80)
        
        exp_dist = self.df['formatted_experience_level'].value_counts()
        exp_pct = (exp_dist / len(self.df) * 100).round(2)
        
        exp_analysis = pd.DataFrame({
            'Count': exp_dist,
            'Percentage': exp_pct
        })
        
        print(exp_analysis)
        self.analysis_results['experience_distribution'] = exp_analysis
        
        return self
        
    def analyze_remote_work(self):
        """Analyze remote work opportunities"""
        print("\n" + "="*80)
        print("REMOTE WORK ANALYSIS")
        print("="*80)
        
        remote_count = self.df['remote_allowed'].value_counts()
        remote_pct = (remote_count / len(self.df) * 100).round(2)
        
        print(f"Remote jobs: {remote_count.get(1, 0):,} ({remote_pct.get(1, 0):.2f}%)")
        print(f"On-site jobs: {remote_count.get(0, 0):,} ({remote_pct.get(0, 0):.2f}%)")
        
        # Salary comparison
        print("\nAverage Salary: Remote vs On-site")
        remote_salary = self.df.groupby('remote_allowed')[['min_salary', 'max_salary']].mean()
        remote_salary.index = ['On-site', 'Remote']
        print(remote_salary.round(2))
        
        self.analysis_results['remote_analysis'] = remote_salary
        
        return self
        
    def analyze_engagement(self):
        """Analyze user engagement metrics"""
        print("\n" + "="*80)
        print("ENGAGEMENT METRICS ANALYSIS")
        print("="*80)
        
        # Overall stats
        print(f"\nViews - Mean: {self.df['views'].mean():.1f}, Median: {self.df['views'].median():.1f}")
        print(f"Applies - Mean: {self.df['applies'].mean():.1f}, Median: {self.df['applies'].median():.1f}")
        print(f"Engagement Rate - Mean: {self.df['engagement_rate'].mean():.2f}%")
        
        # Top positions by views
        print("\nTop 10 Job Titles by Average Views:")
        top_views = self.df.groupby('title')['views'].agg(['mean', 'count']).sort_values('mean', ascending=False).head(10)
        top_views.columns = ['Avg_Views', 'Count']
        print(top_views[top_views['Count'] > 10])  # Filter to positions with >10 postings
        
        # Top positions by applications
        print("\nTop 10 Job Titles by Average Applications:")
        top_applies = self.df.groupby('title')['applies'].agg(['mean', 'count']).sort_values('mean', ascending=False).head(10)
        top_applies.columns = ['Avg_Applies', 'Count']
        print(top_applies[top_applies['Count'] > 10])
        
        self.analysis_results['engagement'] = {
            'top_views': top_views,
            'top_applies': top_applies
        }
        
        return self
        
    def analyze_top_companies(self):
        """Analyze top hiring companies"""
        print("\n" + "="*80)
        print("TOP HIRING COMPANIES")
        print("="*80)
        
        top_companies = self.df['company_name'].value_counts().head(20)
        
        print("\nTop 20 Companies by Job Postings:")
        for idx, (company, count) in enumerate(top_companies.items(), 1):
            pct = count / len(self.df) * 100
            print(f"{idx:2d}. {company:40s} | {count:6,} postings ({pct:.2f}%)")
        
        self.analysis_results['top_companies'] = top_companies
        
        return self
        
    def analyze_locations(self):
        """Analyze geographic distribution"""
        print("\n" + "="*80)
        print("GEOGRAPHIC DISTRIBUTION ANALYSIS")
        print("="*80)
        
        # Country distribution
        print("\nTop Countries by Job Postings:")
        top_countries = self.df['country'].value_counts().head(10)
        print(top_countries)
        
        # Top states (for US jobs)
        us_jobs = self.df[self.df['country'] == 'US'].copy()
        print(f"\nUS Jobs: {len(us_jobs):,} postings")
        
        if len(us_jobs) > 0:
            print("\nTop 15 US States by Job Postings:")
            top_states = us_jobs['state'].value_counts().head(15)
            print(top_states)
        
        self.analysis_results['geographic'] = {
            'countries': top_countries,
            'us_states': top_states if len(us_jobs) > 0 else None
        }
        
        return self
        
    def analyze_applications_types(self):
        """Analyze application process types"""
        print("\n" + "="*80)
        print("APPLICATION TYPE ANALYSIS")
        print("="*80)
        
        app_types = self.df['application_type'].value_counts()
        app_pct = (app_types / len(self.df) * 100).round(2)
        
        print("\nApplication Type Distribution:")
        for app_type, count in app_types.items():
            print(f"{app_type:20s}: {count:8,} ({app_pct[app_type]:.2f}%)")
        
        # Application process ease vs engagement
        print("\nAverage Engagement by Application Type:")
        app_engagement = self.df.groupby('application_type').agg({
            'views': 'mean',
            'applies': 'mean',
            'engagement_rate': 'mean'
        }).round(2)
        print(app_engagement)
        
        self.analysis_results['application_types'] = app_engagement
        
        return self
        
    def analyze_sponsored(self):
        """Analyze sponsored vs organic postings"""
    
        print("\n" + "="*80)
        print("SPONSORED POSTINGS ANALYSIS")
        print("="*80)

     # Get counts
        sponsored_counts = self.df['sponsored'].value_counts()

    # Print counts
        print("\nSponsored Distribution:")
        print(sponsored_counts)

    # Percentages
        sponsored_pct = (sponsored_counts / len(self.df) * 100).round(2)

        organic_count = sponsored_counts.get(0, 0)
        sponsored_count = sponsored_counts.get(1, 0)

        organic_pct = sponsored_pct.get(0, 0)
        sponsored_pct_value = sponsored_pct.get(1, 0)

        print(f"\nOrganic postings:   {organic_count:,} ({organic_pct:.2f}%)")
        print(f"Sponsored postings: {sponsored_count:,} ({sponsored_pct_value:.2f}%)")

    # Engagement comparison
        print("\nAverage Engagement: Organic vs Sponsored")

        sponsored_engagement = self.df.groupby('sponsored').agg({
        'views': 'mean',
        'applies': 'mean',
        'engagement_rate': 'mean'
        }).round(2)

        sponsored_engagement.index = [
        'Organic' if x == 0 else 'Sponsored'
        for x in sponsored_engagement.index
]

        print(sponsored_engagement)
        return self 
        
    def analyze_posting_duration(self):
        """Analyze posting duration patterns"""
        print("\n" + "="*80)
        print("POSTING DURATION ANALYSIS")
        print("="*80)
        
        print(f"\nPosting Duration Statistics (days):")
        print(f"Mean: {self.df['posting_duration_days'].mean():.1f}")
        print(f"Median: {self.df['posting_duration_days'].median():.1f}")
        print(f"Min: {self.df['posting_duration_days'].min()}")
        print(f"Max: {self.df['posting_duration_days'].max()}")
        
        # Duration by work type
        print("\nAverage Posting Duration by Work Type (days):")
        duration_by_work = self.df.groupby('work_type')['posting_duration_days'].mean().sort_values(ascending=False)
        print(duration_by_work.round(1))
        
        self.analysis_results['posting_duration'] = duration_by_work
        
        return self
        
    def generate_key_insights(self):
        """Generate actionable insights for portfolio"""
        print("\n" + "="*80)
        print("KEY INSIGHTS & PORTFOLIO OPPORTUNITIES")
        print("="*80)
        
        insights = []
        
        # Insight 1: Most common job titles
        top_titles = self.df['title'].value_counts().head(5)
        insights.append(f"\n1. TOP JOB TITLES:")
        insights.append(f"   The most posted job titles are: {', '.join(top_titles.index.tolist())}")
        
        # Insight 2: Salary analysis
        avg_salary = (self.df['min_salary'].mean() + self.df['max_salary'].mean()) / 2
        insights.append(f"\n2. AVERAGE SALARY:")
        insights.append(f"   Overall average salary: ${avg_salary:,.0f}/year")
        
        # Insight 3: Remote adoption
        remote_pct = (self.df['remote_allowed'].sum() / len(self.df[self.df['remote_allowed'].notna()]) * 100)
        insights.append(f"\n3. REMOTE WORK ADOPTION:")
        insights.append(f"   {remote_pct:.1f}% of jobs with specified remote status allow remote work")
        
        # Insight 4: Experience level distribution
        entry_pct = (len(self.df[self.df['formatted_experience_level'] == 'Entry Level']) / len(self.df) * 100)
        insights.append(f"\n4. ENTRY-LEVEL OPPORTUNITIES:")
        insights.append(f"   {entry_pct:.1f}% of jobs are entry-level positions")
        
        # Insight 5: Application difficulty
        easy_apply_pct = (len(self.df[self.df['application_type'] == 'External Link']) / len(self.df) * 100)
        insights.append(f"\n5. APPLICATION EASE:")
        insights.append(f"   {easy_apply_pct:.1f}% of jobs use external application links (easier apply)")
        
        for insight in insights:
            print(insight)
        
        return insights
        
    def run_full_eda(self):
        """Execute full exploratory data analysis"""
        print("\n" + "="*80)
        print("STARTING EXPLORATORY DATA ANALYSIS")
        print("="*80)
        
        (self.analyze_missing_values()
         .analyze_salary_distribution()
         .analyze_work_types()
         .analyze_experience_levels()
         .analyze_remote_work()
         .analyze_engagement()
         .analyze_top_companies()
         .analyze_locations()
         .analyze_applications_types()
         .analyze_sponsored()
         .analyze_posting_duration()
         .generate_key_insights())
        
        print("\n" + "="*80)
        print("EDA COMPLETE")
        print("="*80)
        
        return self


# ============================================================================
# VISUALIZATION FUNCTIONS
# ============================================================================

def create_visualizations(df, output_dir='visualizations'):
    """Create key visualizations for portfolio"""
    import os
    os.makedirs(output_dir, exist_ok=True)
    
    # 1. Salary Distribution
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    
    # Histogram
    axes[0, 0].hist(df['max_salary'].dropna(), bins=100, color='steelblue', alpha=0.7)
    axes[0, 0].set_xlabel('Max Salary (USD)')
    axes[0, 0].set_ylabel('Frequency')
    axes[0, 0].set_title('Distribution of Maximum Salaries')
    axes[0, 0].set_xlim(0, 300000)
    
    # By work type
    salary_by_work = df[df['max_salary'].notna()].groupby('work_type')['max_salary'].mean().sort_values()
    salary_by_work.plot(kind='barh', ax=axes[0, 1], color='coral')
    axes[0, 1].set_xlabel('Average Max Salary (USD)')
    axes[0, 1].set_title('Average Salary by Work Type')
    
    # By experience level
    salary_by_exp = df[df['max_salary'].notna()].groupby('formatted_experience_level')['max_salary'].mean().sort_values()
    salary_by_exp.plot(kind='barh', ax=axes[1, 0], color='lightgreen')
    axes[1, 0].set_xlabel('Average Max Salary (USD)')
    axes[1, 0].set_title('Average Salary by Experience Level')
    
    # Box plot
    df.boxplot(column='max_salary', by='work_type', ax=axes[1, 1])
    axes[1, 1].set_xlabel('Work Type')
    axes[1, 1].set_ylabel('Max Salary (USD)')
    axes[1, 1].set_title('Salary Distribution by Work Type')
    plt.suptitle('')  # Remove default title
    
    plt.tight_layout()
    plt.savefig(f'{output_dir}/01_salary_analysis.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # 2. Work Type Distribution
    fig, axes = plt.subplots(1, 2, figsize=(15, 6))
    
    work_counts = df['work_type'].value_counts()
    work_counts.plot(kind='bar', ax=axes[0], color='steelblue')
    axes[0].set_xlabel('Work Type')
    axes[0].set_ylabel('Count')
    axes[0].set_title('Job Postings by Work Type')
    axes[0].tick_params(axis='x', rotation=45)
    
    work_pct = (work_counts / len(df) * 100)
    axes[1].pie(work_counts, labels=work_pct.index, autopct='%1.1f%%', startangle=90)
    axes[1].set_title('Percentage Distribution of Work Types')
    
    plt.tight_layout()
    plt.savefig(f'{output_dir}/02_work_type_distribution.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # 3. Experience Level Distribution
    fig, axes = plt.subplots(1, 2, figsize=(15, 6))
    
    exp_counts = df['formatted_experience_level'].value_counts()
    exp_counts.plot(kind='bar', ax=axes[0], color='coral')
    axes[0].set_xlabel('Experience Level')
    axes[0].set_ylabel('Count')
    axes[0].set_title('Job Postings by Experience Level')
    axes[0].tick_params(axis='x', rotation=45)
    
    exp_pct = (exp_counts / len(df) * 100)
    axes[1].pie(exp_counts, labels=exp_pct.index, autopct='%1.1f%%', startangle=90)
    axes[1].set_title('Percentage Distribution of Experience Levels')
    
    plt.tight_layout()
    plt.savefig(f'{output_dir}/03_experience_distribution.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # 4. Engagement Metrics
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    
    # Views distribution
    axes[0, 0].hist(df['views'].dropna(), bins=100, color='steelblue', alpha=0.7)
    axes[0, 0].set_xlabel('Views')
    axes[0, 0].set_ylabel('Frequency')
    axes[0, 0].set_title('Distribution of Job Views')
    axes[0, 0].set_xlim(0, 100)
    
    # Applications distribution
    axes[0, 1].hist(df['applies'].dropna(), bins=100, color='coral', alpha=0.7)
    axes[0, 1].set_xlabel('Applications')
    axes[0, 1].set_ylabel('Frequency')
    axes[0, 1].set_title('Distribution of Applications')
    axes[0, 1].set_xlim(0, 50)
    
    # Top companies
    top_companies = df['company_name'].value_counts().head(10)
    top_companies.plot(kind='barh', ax=axes[1, 0], color='lightgreen')
    axes[1, 0].set_xlabel('Number of Postings')
    axes[1, 0].set_title('Top 10 Hiring Companies')
    
    # Remote vs On-site engagement
    remote_data = df.groupby('remote_allowed')[['views', 'applies']].mean()
    remote_data.index = ['On-site', 'Remote']
    remote_data.plot(kind='bar', ax=axes[1, 1], color=['steelblue', 'coral'])
    axes[1, 1].set_xlabel('Job Type')
    axes[1, 1].set_ylabel('Average Count')
    axes[1, 1].set_title('Average Engagement: Remote vs On-site')
    axes[1, 1].tick_params(axis='x', rotation=0)
    
    plt.tight_layout()
    plt.savefig(f'{output_dir}/04_engagement_analysis.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    print("✓ Visualizations saved to 'visualizations/' directory")


# ============================================================================
# USAGE EXAMPLE
# ============================================================================

if __name__ == "__main__":
    # Load cleaned dataset
    df = pd.read_csv('postings_cleaned.csv')
    
    # Run EDA
    eda = JobPostingsEDA(df)
    eda.run_full_eda()
    
    # Create visualizations
    create_visualizations(df)
    
    print("\n✓ EDA complete! Check visualizations/ folder for charts.")
```
